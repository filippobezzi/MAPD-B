# Data Management

In the following discussion we consider the steps that are taken from after the data is acquired to the production of useful information. 
1. Data Acquisition Sysyem: data is acquired using a system made up of the following components:
+ Sensors are used to convert physical measurements in electric signals.
+ Frontend electronics act as an interface between the sensors and the system that elaborates the signal by amplifying, discriminating and digitizing the signal.
+ Readout electronics componenents are responsible for the digital conversion of the analogic signal (can also act as a temporary memory for the signal).
+ Trigger & DAQ decide which events are interesting enough to keep and the final dat is collected and aggregated from all sensors into a structured form.
2. Computing Model: actions are implemented on the data using the following components:
+ Storage: the modality and the location in which data is stored (form data to datasets, file-systems, ...).
+ Reliability and Security: The means and strategies of preserving data and mantaining it reliable (authentication, authorization, ...).
+ Querying and Filtering: The means and techniques of accessing data (data-mining, feature-enrichment, ...).
+ Processing: The implementation of data in algorithms that gain useful information and obtain interpretable results (higher-level computation, training and testing of algorithms).
Sources of data are diverse and it can be acquire using sensors (experiments, IoT, ...), generative simulations (MontCarlo algorithms), Text (log-files, docs), and Heterogeneous sources (GPS location + image + ...).

A dataset is a collection of data, a set of information collected by a number of sources thoughout a stretch of time associated to a unique field or purpose.

Metadata are a form of data containing information needed to interpret the data and allow to access, describe or process the dataset correctly.

There are different types of data:
+ Structured: containing a defined data type, format and structure (data schemas are knwon and fixed). An example are .csv files.
+ Semi-structured: The logic with which they are formed are structured and possess a flexible schema. Examples are .json files. Format and structure can be extracted by parsing the data. By defininf a schema it is possible to recast the data to structured data formats.
+ Unsctructured: data that belongs to no fixed schema or sturcture, they are defined by a continuous flux of information (such as text, audio, video, or image data).

Data management deals with all the needs and challenges related to the (safe)keeping and accessibility of datasets. Such tasks can be split in a varyety of subset tasks:
+ Data storage
+ Reliability & Long-term preservation
+ Accessibility & security
+ File/Database management systems
+ ...

## 2. Data Storage

+ DAQ writes in a memory with at very hih frequencies.
+ Data is processed every few hours / days, then reprocessed every ~year, in-memory writing occurs less frequently.
+ Larger storage units are used for long term storage, data is written in the memory once.

Then a growing number of thoughput and I/O operations correspond to a lower storage utilized and vice-versa.

Useful terms:
+ Latency: the time needed to execute a single I/O operation.
+ Frequency: the rate at which I/O operations are exectued.
+ Block size: The dimension in byte of a single I/O operation.
+ Throughput (Bandwidth): The measured (theoretical) capacity of data transfer over time. It is expressed as the product between the block size and the frequency, that is how many bits are transferred in some time interval (i.e. $MB/s$).

Example:
+ IO --> Write operation
+ Block size --> 4kB
+ Total transfer size --> 1000 $\times$ 4kB = 4MB
+ Measured latency = 11.1 ms = 0.0111
We can find the following quantities:
+ Estimated throughput = Total transfer size / Measured latency = 360 MB/s
+ Estimated IOPS (I/O per second) = # Transfers / Measured latency = 1000 / 0.0111 = 90_000

### Storage Devices

#### DRAM (Dynamic Random  Access Memory)

Stores data bits in NAND cells. RAM means that data can be access from any location at the same speed. However, when the device is turned off the data is lost (temporary storage, low latency).

#### NVM (Non-Volatile Memory)

Very expensive alternative to the DRAM (non-volatile means that data is not lost when flow of current stops).

#### SSD (Solid State Drives)

Data is stored in a *multi-levle charge trap*: capacitors distinguish between more levels (not only 1/0 like NAND ports), usually 8, that is three bits of information. Data is stored by trapping electrons in cells. Information is then obtained by measuring the threshold voltage required to change the cells state (that is bits are encoded in each cell in the number of electrons stored).
The TLC (*Triple-level cell*) is able to store 3 bits per cell by differentiating 8 distinct charge levels (0, 1, 2, 3, ..., 7 electrons stored). 
This multi-levle charge trap technology guarantees a greater *storage density* at the cost of latency for writing data (reading $~10 \mu s$ , writing $~100 \mu s$). That is because reading involves only measuring the voltage of a cell, whereas writing requires clearing the cell and precisely setting the charge level. 
Furthermore, each writing degrades the cell slightly being subject to an inevitable process of hysteresis, as residual charge eventually comprosimes the effectiveness of the encoding. In order to avoid error, one can only avoid to write in cells that have exceeded a maximum number of writing cycles.

#### HDD (Hard Disks Drives)

A series of double-sided spinning *platters* (disks) covered in magnetic media. Each magnetic domain is called *track*, with data being stored in contiguou *sectors* of the platter tracks. The reading of the data then occurs per *sector* (see image in slides).
I/O requests are managed by a moving arm assembly that places its head on the appropriate track to R/W data.
Latency ($~10 ms) generally depends on a number of factors: the time to move the head to the proper track, the time for the disk to rotate and align to the appropriate sector.

#### Magnetic Tapes

Data is stored sequentially (contiguous) over the tape such that the physical support (the tape) can be rolled up in a compact form. An I/O request generally requires much longer times as: a robotic arm fetches the cartridge, the tape needs to be unrolled to the point where the data starts, all storage locations must be read sequentially.
It makes sense that such data storage devices are used for write-once operations and long term safekeeping.


### Kryder's Law

Naturally one wonders how much innovation in the hardware technology can lead to actual advancements relate to better ratios in the latency and capacity of storage devices.
Kryder's law (2005) says that *“The storage density (amount of data stored per unit of disk/chip area) doubles every 13 months”*, meaning that the cost of storage devices decreases quickly over time. However Kryder's law hasn't been valid for the last two decades, indicating that we are reaching a plateau of technological advancement in the current framework.

### Scalability

In order to handle an increase in the size of dataset, there are different ways to increasing storage capacity and mantain or optimize the R/W performance.
One can scale the storage device system vertically (getting better / more performant / higher capacity hardware devices), or horizontally (chaining together more seources, and create a distributed system).
The vertical approach
- has the *pros* of not brining any significant change into how the users access the resources, which usually can be accessed quickly (meaning directly or simply with less latency if one gets a better device), 
- the *cons* are that it is usually expensive, improvements are limited, and operations need to be stopped for the time required to update once's system.
The horizonal approach 
- has the *pros* of allowing one to create very large systems (less expensive) that can be implemented keeping the necessary parts of the system running.
- The *cons* are that when multiple hardware resources are connected together the network bandwith can be a limiting factor and solutions to access the data are more complex.



## 3. Reliability and Security

Once the data is stored it must be preserved, in any way possible. It is not important to preserve the physical device as long as one ensures that if something happens the data contained in it is safe. One must protect stored data from the following threats:
1. Data loss due to storage failure (HW failure, electrica / mechanical / ...), usually all data is lost.
2. Data corruption, which causes are single-bit errors for a range of I/O operations. All of some data can be recovered. (HOW?)
3. Software / data-transfer errors, when consistency checks are not performed when copying or transfering data.

**Reliability** corresponds to the probability that a system continuous to operate correctly over a certain amount of time or a certain number of operations.

Quantitatively one defines the **MTTF** (Mean Time To Failure), that is the average time that passes from the moment the device is turned on and the moment failure occurs. For reparable systems the **MTBF** (Mean Time Between Failure) expresses the average time thath passes between each failure.

In reality:
- MTTF of HDD is ~10 years
- the probability tha that the device will break daily is $p = \frac{1}{10 \cdot 365} = 0.00027$ ("failures per day")
- using 5000 HDD means $5000 \cdot 0.00027 = 1.35$ failures per day.

### Reliability Strategies

Data storage reliability can be improved with a number of strategies:
+ **Mirroring**: Replicate data across multiple storage elements (and sites).
+ **Striping**: Subdivide data in "pieces" and scatter it across multiple storage elements.
+ **Checksum / Partity checks**: Ensure data consistency by detecting (and possibly correcting) errors.

The strategy **disk redundancy**, indicated by the acronym **RAID** (Redundant Array of Indipendent Disks), is instead used to improve storage reliability. Simply more *physical* devices are combined together to form a single *logical* one. The different ways data can be stored in RAIDS gives different schemas, so one sata that different RAID schemas (levels) can be implemented to store data (RAID 0, RAID 1, RAID 1+0, RAID 5, RAID 6, many others..).

#### RAID 0 -> Disk Striping, No Mirroring

+ Use 2+ disks across data, data is striped across all the disks, and isn't mirrored (i.e. no replication).
+ $N$-disks combined give $N\times$-capacity of smallest disk.

Such set up can be used for performance (can R/W at the same time from all disks), but not suitable when high reliability is required.

#### RAID 1 -> Disk Mirroring, No Striping

+ Use 2+ disks to store data, which is mirrored, but no striped (no data subdivision).
+ Overall storage capacity 1$\times$ the capacity of the smallest one.
+ Highly rerliable at the cost of performance, as one writes at the slowest throughput.

#### RAID 1+0 (or RAID 10) -> Disk Mirroring + Striping

+ Use 4+ disks, data is mirrored across 2 disks (RAID 1) and striped across 2+ RAID 1 disks.
+ Good overall performance and improved reliability (R faster due to striping / W slower due to mirroring).

#### RAID w/ Parity

A **parity** is information that can be stored in deicated data blocks as an error protection scheme (usually based of the **XOR** logic). Consider we have three data blocks D1, D2, and D3, in case a single disk fails, data can be recovered from a single parity bit information. However, in case of a double disk failure, data cannot be recovered. 

#### RAID 5 -> Disk Striping, No Mirroring, w/ Distributed Parity

+ Use 2+ disks to store data, +1 to store parity.
+ Data is striped across all the disks, no mirroring.
+ Parity block is distributed across all disks.

$$
\begin{equation*}
p = D_1 \oplus D_2 \oplus D_3
\end{equation*}
$$

then, say $D_3$ is lost

$$
\begin{equation*}
D_1 \oplus D_2 \oplus p = D_3
\end{equation*}
$$

Such framework (similar to RAID 0) gives better performances guaranteeing recovery of single disk / data loss. However, a greater throughput is required as in both R/W operations the computational cost of the calculation of the partity for all data bits is not trivial.

#### RAID 6 -> Disk Striping, No Mirroring, w/ Distributed Double Parity

+ 2+ disks to store data
    + Data striping across all disks
    + No mirroring
+ +2 to store paratiy (two parity blocks distributed across all disks)

Data can be recovered even in the case of double error or loss of two disks, as long as the bits of information are linearly indipendent. That can be done using "error-correcting" codes such as the Reed-Solomon codes.
$$
\begin{equation*}
p = D_1 \oplus D_2 \oplus D_3
\end{equation*}
$$

$$
\begin{equation*}
q = D_1 \oplus (D_2 \gg shift) \oplus (D_2 \gg shift^2)
\end{equation*}
$$

If the two parity words carried the same information the system would collapse to a single equation and two unknowns. 

### Security

The role and objective of security strategies in data management is essentially to ensure:
- **Confidentiality**: Nobody can extract knwoledge of what you transfer (even when "listening" -> "masking").
- **Integrity**: Messages are received as they were sent, no modification has occured.
- **Aunthenticity, Identity, Non-repudiation**:
    - One can verify the exchange of information is happening with the intended party,
    - and who is the specific indibidual behind that entity,
    - and the individual behind that asset cannot deny being associated with it.

**Cripytography** is as the basis of all modern security techniques. It is defined as:

*The discipline that embodies the principles, means, and methods for the **transformation of data** in order to **hide their semantic content**, prevent their unauthorized use, or prevent their undetected modification.*

#### Authentication & Authorization

**Authentication** refers to possibility of identifying users and validating their identity by means of:
- Public/private keys
- Certificates
- Single-Sign-On (SSO)
- Custom authentication
- other ...

**Authorization** refers to grating a user permission to access/use specific resources.

Usually authentication is done through an *authorization server*, which certifies the connection between a specific user and a tool or service. After authentication has been done the sistem verifies permissions of the user through an *Access Control List*. After the authentication and authorization steps are complited the user is able to execute the desired operation. The user is *responsible* for the outcome of that operation.

#### Encoding/Deconding Information

+ Symmentric Key Encryption: 

    Essentially a key K is used to scramble (cypher) and reassemble (decypher) the data. A logic function is used to encode-decode the information. So if a message is expressed as a series of bits, then the key must but a series of bits of the same length. This simple method of enctryption requires that the key is exchanged physically, making it highly inconvenient.

+ Asymmetric Key Encription:

    The most widely used encription method. Two keys are used:
    + one is *public*: it is used only to **encrypt** the data,
    + and one is *private*: used to **decrypt** the data.

    The algorithm produces the two keys by applying a one-way function that is easy to run in one direciton (the public key) and practically impossible to reverse unless one knows a hidden "trap-door" (the private key). The transformation of the message created using the public key produces an output that can be inverted only if one known the mathematical secret linked to that public key. That is because the public key is generated by applying a one-way function to the private key, such that recovering the private key is computationally infeasible. 
    
    See the following example scenario:
    1.	User B generates (public, private) key pair and publishes the public key.
	2.	User A retrieves B’s public key and computes the ciphertext with it.
	3.	The ciphertext flows over the network. An attacker sees only (ciphertext, public key).
	4.	User B applies the private key to invert the one-way function and obtains the plaintext.
	5.	Everyone else would need to break the underlying hard problem to do the same, which is considered infeasible.

     

#### Cryptographic Hash Functions

**Hashing** algorithms are designed to scramble (transform) the data into a fixed size version of it, called **disgest**, in such a way that any small difference in the original data will result in a totally different digest. The case in which two different datasets give the same digest is called **hash collision**. Hashing algorithms are then designed to avoid hash collisions. This characteristic of hash functions makes them useful in:
- Masking/Protecting data: sensible data can be saved as a digest instead of plain text, then only the same data will correspond to identical digests and this can be used for authentication .
- Data integrity verifications: again, check if digest of copied data is identical to the original one to gain insight into whether data was copied correctly
- Digital signature verification...

REMARK: Encryption vs. Hashing

+ Encrytpion: 
    - **Encode and compress** data for confidentiality
    - Must be **fully reversible** to allow decryption
    - Can be computationally expensive

+ Hashing algorithms:
    - scramble data into fixed-size digest to **hide/check** data
    - **not reversible**
    - produce completely different digest for any small difference in data
    - computationally efficient and **avoid hash collision**.

## 4. File Systems

Datasets can be stored and/or managed using:
+ **Files**: Blocks of data are stored in volumes, managed by system, namely a *File System*, which can be extended to  a number of memory units (horizontal scaling) to make a *Distributed File Systems*.
+ **DataBases**: Data is stored and managed via dedicated systems, such that every single element is accessible at a *granular* level and modifiable through systems called *DataBase Magnagement Systems*, which can be extended to *Distributed DBs*.
+ **ObjectStores**: Data is saved in arbitrary formats (files, blocks, operating systems) stored in *buckets*, which are used in *Cloud Storage* platforms.

The FS manages and tracks the location of data in terms of a **hierarchical** structure, using *directories*. The characteristics of a File System are essentially the following:
+ **Storage of data** in data blocks, scattered within the available storage devices.
+ **Abstraction** for identificatio of data as files/directories by their name, path, and metadata.
+ Interface with the Operating System (OS) for allowing **low level operations**.

### File Systems Types
Depending on the File System, files can be accessed, and disk space allocated, in a number of ways, including:

+ **File Access**:
    1. Sequential Access
    2. Direct/Random Access
    3. Indexed Access
+ **Space Allocation**:
    1. Contiguous Allocation: 
    Blocks that constitute the files are physically saved one next to the other, meaning access to the file is of sequential type. The limitating factor is that once the file is stored, if one can't increase the file size beyond a certain threshold unless a block containing data from another file has to be overwritten. (EX: Magnetic tapes, used when data does not need to be modified)
    2. Linked Allocation:
    Space allocation method in which a file is stored in blocks together with the information of where the successive data block is found. In this sense blocks are linked toghether and files can grow indefinitely at the cost of a fraction of memory required to store the linking information.
    3. Indexed Allocation:
    An indexed FS uses *direct* and *undirect* blocks. The first block stores: the files *metadata*, the direct blocks containing the indices with the information of the *data blocks* in which the file's raw data is stored, or (single-,double-, triple-, ...)*indirect blocks* that *point* to other data blocks or pointers.

#### Indexed FS Example

We consider a indexed file system made of:
- 12 direct blocks (point to data)
- 1 indirect block (point to pointers)
- Each pointer is 4B
- Each direct data block contains 4kB of data

Then each of the twelve direct blocks contains data a total of

\begin{equation*}

$$.
The indirect block contains only indices, than it can contain the following number of indices (pointers):
$$
\begin{equation*}

$$,

$$
\begin{equation*}

In the case of a double indirect block, of equal storage capacity and containing indices of the same size the total storage capacity of such file system would be:

\begin{equation*}

\end{equation*}
$$.

#### Limiting Factors of FS

From the above example we see that the main limiting factor of file systems is that, depending on their structure, it will be characterized by a maximum file size it can store. Furthermore, it can be complex to define a good data hierarchy for large datasets. Also as files acess is provided with an "atomic view" no underlying structure is accessible, and it is not possible to group-select-filter the data based on file content.

### Distributed File Systems (DFS)

As all data storage systems, FS can be scaled vertically or horizontally. In principle, in the case of vertical scaling the file system interfaces directly with a machine that provides more memory. In the case of horizontal scaling, the file system is extended to environments in which both data and processing is spread out across several *nodes*, which are the machines with attached storage. Such file systems are called Distributed File Systems (DFS). The *cluster* is the set of computing resources, that is of nodes (computers), that make the DFS. A user connects though a personal computer through a program called *client*. The nodes are connected between them though a network called *communication network*.

#### How DFS Work

Depending on the user type and preferred hierarchy for data storage and management DFS can be different. Some examples are:
+ **NFS** (*Network File System*): OS UNIX, most used for user-friendly OS.
+ **AFS** (*Andrew File System*): academic and research oriented OS.
+ **Lustre**: specifically for HPC (*High Performance Computing*) in order to maximise the throughput and manage a lot of data.
+ **EOS** (*Elastic Object Storage*): developed by CERN to manage a large amount of unstructured data.

Differently from local FS:
1. Block sizes are distributed on different machines, their location is not only specified by the data block index by also by the machine (node) index.
2. Latency is increased by that related to the implementatio nof the communication network, which is much greater than a local file system.
3. Because of this latency issue block sizes are usually greater, in order to maximize the amount of data extracted from each block location.

#### Ideal DFS Requirements

DFS's are developed trying to achieve the following ideal requirements:
1. **Transparency**:
    Users should be a able to access the system regardless where they log-in, perform the same operations on DFSs and local FSs, meaning also not experience potential failures of part of the system.
2. **Fault Tolerance**:
    Distributed system should not halt in case of failure of parts of its infrastructure.
3. **Scalability**:
    The system should efficiently leverage from the usage of large numbers of severs (potentially dynamic) added into the system (i.e. be extensible with the increase of storage nodes).

#### HDFS

The **HDFS** (*Hadoop Distributed File System*) is an open-source DFS realize in 2006 in the attempt of emulating the GFS (*Google File System*). It was developed with the goal of managing and working with a large amount of data on *commodity hardware*. The system is structure as follows:
+ The 1 **namenode** contains the metadata of all the files and has the role of accepting or refusing the operations that are sent to users. 
+ The many **datanodes** host the data blocks (typical block size is 128MB). 
Each operation mainly follows these four steps:
1. The client issues the request (R/W or other) to the namenode.
2. The namenode checks for authentication/access priviledge/etc and provides the locaton where to write the data blocks.
3. The client starts writing the data blocks directly to the datanodes w/o passing through the namenode
4. The datanodes report back to the client to acknowledge the write completion.

Regarding the three main requirements of DFS, HDFS effectively satisfies the Transparency and Scalability requirements, as a user can operate directly from a personal computer through the client, and, by design, a new node can be easily added to the network of datanodes already present. Regarding fault tolerance, there are three main issues:
1. **Network instability**: Part of the network links may fail meaning operations one or more datanodes cannot take place. 

    A possible solution is to provide the datanode with the function of repeatedly sends a sort of "heartbeat" signal to the namenode. When a datanode is (even temporarily) non-recheable it will be excluded from operations by the namenode.
2. **Datanode failure**: A datanode fails, meaning data inside it is lost.

    A possible solution is that data blocks are replicated across different nodes (typically three copies of the same block are distributed in three different nodes).
3. **Namenode failure**: The namenode fails, meaning access to metadata and interface with the client is lost. This is an example of SPOF (*Single Point Of Failure*), all information about how to recompose the file from the data blocks is lost.

    A solution is to take a backup copy of the namenode. When the namenode is substituted the whole system stops.

#### CAP Theorem

In computer science three concepts, typically referred to as the CAP theorem, are crucial when dealing with distributed systems (both for data storage and processing):

1. **Consistency**: Every request made to any data node must return the very same copy of the data, and the most recent version.

    As data blocks containing the same data are copied on different datanodes *consistency* refers to the fact that that information needs to be equivalent at all times.
    This is assessed by introduction a "waiting line" in which the user is placed after a file is modified (or an operation is completed, onlt after the copies of that data have been updated the client can proceed with a new operation).

2. **Availability**: Every request (access/W/R/update data) by every client must give an *immediate* response, *indipendently* on the state of the data and of the mal-functioning of some nodes in the network.

    By dispersing effectively the the data blocks in the various nodes, more users can access the same file, even if some datanodes fail. Furthermore, even if the connection between datanodes fails, one wants all system's datanodes to respond ASAP.

3. **Partition Tolerance**: The system must continue to operate despite failures in the communication between nodes (when the communication fails each isolated set of nodes is called a *network partition*).

    So in the case of the formation of network partitions it is necessary that at least one datanode containing the desired data block is always available to the client.

Clearly it is **impossible for a distributed system to simultaneously provide more than two out of the three C/A/P guarantess**, which is exactly what the CAP theorem states. As in distributed systems it is highly improbable that network partitions do not occur, one should choose between a system with a greater degree of Consistency or Availability:
+ **AP Systems**: A system such data data can be accessed any time, but there is no guarantee that the data will contain the most recent value for each read.
+ **CP Systems**: A system such that one reads the most recent value of the data at all times, however access to the data will not always be available as there can be waiting times.
+ **AC Systems**: Common choice for vertically scalable systems, so systems in which network partition does not occur or is not an issue.

## 5. DataBases

A **database** is an ensemble of (generally related) *records* organized and processed by a dedicated *management system*.

Usually the term *database* encapsulate these three concepts:
1. the *records*
2. the *collections* that group single *records* together
3. The *relationships* that exist among records belonging to the same collection or to different collections.

The **database management system** (DBMS) is a software that provides a number of basic functionalisties to operate on the DB, including:
- defining / creating data
- querying / selecting data
- manipulating / processing data.

DBs differentiate from FS in the management of large datasets in the following ways:

- The scale on which the two softwares operate is different. DBMSs work with the granularity of individual records, meaning one can operate on single events. (D)FSs operate only on the while file.
- Data management in FSs is done among different files and directories, whereas in DBMSs within one single entity (the DB) in which are included also the relations between all the data.
- In order to compute operations on the data in FSs one overwrites a copy of the updated file, whereas in DBMSs it is possible to select only a subsection of the data from the data without necessity of replicating them and occupying more memory.

### Database Models
+ A DB **model** (or DB **schema**) refers to the description of the data and the relationships across various parts of the dataset.
+ A DB **instance** (or DB **state**) refers to the actual data contained in the database at a given moment in time.

Not all kinds of data can be described according to a simple format logic schema. It is the with the DBMS that one defines the schema and updates/manipulates instances of a DB.

Think of each DB records as a data **VALUE** indexed by a **KEY**, used as a unique data identifier making the record *recheable*.

In general one distinguishes between two DB models: *relational* and *non-relational*. The former is used for structured data and are also called *SQL database*. The latter is typically developed for semi- or un-structured data types, and are referred to as *NoSQL* ("Not Only SQL") DBs.

#### Relational DBs

In relational DBs one can make the following distinctions:
+ The *records* correspond to the single rows of a table (also called *tuples*).
+ The **attributes** refer to the single columns.
+ The **primary key** is a specific (or combination) attribute *unique* for each record and *minimal*, meaning that the minimum number of attributes is used to define a key.

To link more relatable table one uses a **foreign key**, that is an attribute or a combination of attributes that work as primary key in another table.

An ensamble of operations executed on a relational databse is called a **transaction**, and any of them must satisfy these four criteria:
+ **Atomocity**: (*all-or-nothing*) All the operations in a transaction must be completely executed, otherwise the transaction is aborted.
+ **Consistency**: Any transaction must lead the DB from a *valid state to another valid state*, meaning that all DB constraints are conserved.
+ **Isolation**: Every transaction must generate the same results, as in a single-user environment, and must be indipendent of other transactions, meaning they will lead to coherent results among all users.
+ **Durability**: After a transaction, the DB state persists until it is changed by another transaction.

#### SQL

SQL is a declarative language which contains the features of:
- **Data Definition Language** (DDL): Define DB schemas.
- **Data Manipulation Language** (DML): Query and modify data.
- **Data Control Language** (DCL): Perform authorization and access control.

2. The **constraints** refer to a characteristic the attribute must have, such as `**PRIMARY KEY**`, `**NOT NULL UNIQUE**`, `**DEFAULT <value>**`, `**CHECK (OtherAttr>0)**` (which gives error if a condition is not satisfied).

3. **Alteration** operations involve all the records included in the table. The process of the defininf and maintaning the schema of an RDB is referred to as **Database normalization**.

    + Add a contraint to a pre-existing atttribute:

#### DB Partition

**Partitioning** refers to the action of subdividing the DB into multipla smaller parts. One talks about *vertical paritiioning* when separating table by attributes columns, and *horizontal partitioning* (or *sharding*) when subdividing by whole records. 

The former partitioning methods cna be correctly implemented in RDBMSs as each partition can be treated as different tables with common primary key, unlike the sharding operation which splits the table making it impossible to operate on it directly 

+ RDBMSs require that each table be entirely saved in a single memory element as in case a table is physically scattered one would need a *global* catalogue that knows on which node every row is stored.

Sharding however is preferred as records remain "whole". One can develop a new type of database that can keep track of the distribution of the records. Consider the primary key being the attribute that distinguishes each of the records. **Consistent hashing** consists in mapping the digest (resulting from a hashing function applied to the primary key) in a *ring of values* that goes from $0$ to $2^{\# digest \ bits}-1$. One assigns to each storage unit one or more *token* values. Each token divides the ring and defines a **superior** limit of the "sector" which data the node stores, such that a record is allocated in a node which token is the first between token values greater than the digest. When a new node is added, one can simply assign to it a token and move the records immediately below that value to it.


#### DB NoSQL

**N**ot-**O**nly-**SQL** DBs were developed from the demand of a DBMS that does not require the use of SQL language and is able to manage unstructured and semi-structured data. The difference btween traditional DBs are essentially the following:
1. **Flexible data models**:
    Allow management of databases containing semi- and non-structured data.
2. **Distributed horizontal scalability**:
    Allow to scale-out horizontally with commodity hardware.
3. **Simpler queries**:
    Languages that don't use complex operations a `JOIN`s, prioritizing more rapid response times w.r.t. the expressive power of the language.
4. **Looser consistency models**:
    Not complicancy with ACID allowed, in favor of scalability and *availability* necessary in distributed systems.

The transaction model mostly used is the **BASE**, which stand for:
+ **Basically Available**: The single nodes in a network are generally accessible *almost* alwats (in ACID model access can be interrupted due to an R/W operaiton taking place in that moment due to *isolation*)
+ **Soft state**: The state of a database is not "rigidly" *persistent*, there exists the possibility that it changes without specific intervention of the user, as in the case of the replication of the data on the various nodes of a cluster.
+ **Eventually consistent**: The *consistency* is guaranteed only beyond some time and not necessarily on the exact moment the transaction is concluded.

##### Key-value DBs

Simplest data storage method is used: to each record is associated a key (similarly to dictionaries). It is *schema-less* as data can be stored in any format and one does not need to include metadata (there is not a data structure to specify, usually one would expect that belonging to the same "data gathering operation" belongs to a set of records, each record containing a feature (key) with an associated value. This key-value DB model works very well in cases that require a large number of R/W in rapid succession. In such cases data is saved directly on the RAM, such databases, that manage data stored on the primary memory, are named **in-memory databases**.

##### Document DBs

It is a document oriented DB in which entire *documents* are stored. Documents are made of field-value pairs (essentially dictionaries with information econdoded in JSON format) and are organized in "collections" through a field called `id`, which identifies uniquely the document itself (like a primary key). Queries are executed using the field of the documents.

##### Wide-Column DBs

Again a key-value structure is used to store data, however each record is identified by a "row-key" and values contain an arbitrary number of columns (even not a column but yes columns before and after).

##### Time series DBs

Developed to memorize data indexed by timestamps and to group them in time windows. In particular they work very well with empirical data produced by the sensors and providing functionalities to calculate metrics and statistics that help identify trends and patterns.

##### Graph DBs

DBs designed to store as records the relationships between nodes, rather than the nodes.

##### Vector DBs

Records are vector data representation in a high dimensionality phase-space. The most notable example is the databse to store the vector embeddings of language models. In this example, each record carries within it semantic information. These DBs are optimized for similarity searches: search the most similar term to a word once indexed as a vector.
